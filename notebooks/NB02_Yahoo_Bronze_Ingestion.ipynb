```python
# ============================================================
# NB02_Yahoo_Bronze_Ingestion
# Purpose:
#   Pull current available data directly from Yahoo Finance
#   through yfinance for the Lynch Slow Grower / Stalwart project.
#
# Reads:
#   bronze_universe
#   Files/config/lynch_parameters.csv
#
# Writes:
#   bronze_yahoo_prices
#   bronze_yahoo_dividends
#   bronze_yahoo_fundamentals_long
#   bronze_yahoo_market_snapshot
#   bronze_yahoo_ingestion_log
#
# Notes:
#   - Initial run downloads price history from configured start date.
#   - Later runs download only recent price dates with an overlap.
#   - Fundamentals are overwritten only if sufficient Yahoo responses succeed.
#   - No inline %pip installation is used.
# ============================================================

import time
import uuid
import numpy as np
import pandas as pd
import yfinance as yf

from datetime import datetime, timedelta
from delta.tables import DeltaTable

from pyspark.sql import functions as F
from pyspark.sql.types import (
    StructType, StructField,
    StringType, DateType, DoubleType,
    LongType, TimestampType
)

# ------------------------------------------------------------
# 0. Run metadata and configuration
# ------------------------------------------------------------

RUN_ID = str(uuid.uuid4())
RUN_TS = datetime.utcnow()
TODAY = datetime.utcnow().date()

DEFAULT_PRICE_START = "2005-01-01"
DEFAULT_OVERLAP_DAYS = 10
SLEEP_SECONDS = 1.0
MIN_FUNDAMENTAL_SUCCESS_RATIO = 0.60

print(f"RUN_ID: {RUN_ID}")
print(f"RUN TIME UTC: {RUN_TS}")

# ------------------------------------------------------------
# 1. Helpers
# ------------------------------------------------------------

def table_exists(table_name):
    """Return True if Lakehouse table exists."""
    try:
        spark.read.table(table_name)
        return True
    except Exception:
        return False


def safe_float(value, default=None):
    """Convert a value to float, returning default for null/NaN/infinity."""
    try:
        result = float(value)
        if np.isnan(result) or np.isinf(result):
            return default
        return result
    except (TypeError, ValueError):
        return default


def to_python_date(value):
    """Convert pandas/Python date-like value to datetime.date."""
    try:
        return pd.to_datetime(value).date()
    except Exception:
        return None


def get_fast_info_value(fast_info, key):
    """
    FastInfo can behave like a mapping but may raise errors for missing fields.
    """
    try:
        if hasattr(fast_info, "get"):
            return fast_info.get(key)
        return getattr(fast_info, key, None)
    except Exception:
        return None


def flatten_statement(statement_df, symbol, yahoo_symbol, period_type, statement_type):
    """
    Converts a Yahoo statement DataFrame to long format.

    Output:
      symbol, yahoo_symbol, period_type, statement_type,
      period_end, source_field, value, ingested_at, run_id
    """
    records = []

    if statement_df is None or statement_df.empty:
        return records

    # Iterate row by row. Yahoo uses row names as source fields
    # and columns as statement reporting dates.
    for source_field, values in statement_df.iterrows():

        # Some Yahoo frames may contain duplicate source fields.
        # Keep source field name and each available period/value.
        for period_end, value in values.items():
            metric_value = safe_float(value)

            if metric_value is None:
                continue

            report_date = to_python_date(period_end)
            if report_date is None:
                continue

            records.append((
                str(symbol),
                str(yahoo_symbol),
                str(period_type),
                str(statement_type),
                report_date,
                str(source_field),
                metric_value,
                RUN_TS,
                RUN_ID
            ))

    return records


def get_start_date_by_symbol(universe_df, table_name, global_start, overlap_days):
    """
    If price table exists, fetch from latest known date per symbol minus overlap.
    Otherwise fetch from global historical start date.
    """
    start_map = {
        str(row["symbol"]): global_start
        for _, row in universe_df.iterrows()
    }

    if not table_exists(table_name):
        return start_map

    existing = (
        spark.read.table(table_name)
        .groupBy("symbol")
        .agg(F.max("trade_date").alias("last_trade_date"))
        .toPandas()
    )

    for _, row in existing.iterrows():
        symbol = str(row["symbol"])
        last_date = row["last_trade_date"]

        if symbol not in start_map or pd.isna(last_date):
            continue

        last_date = pd.to_datetime(last_date).date()
        incremental_start = last_date - timedelta(days=overlap_days)

        # Do not go before full history start.
        start_map[symbol] = max(global_start, incremental_start)

    return start_map


# ------------------------------------------------------------
# 2. Read parameter file
# ------------------------------------------------------------

try:
    params = (
        spark.read.format("csv")
        .option("header", "true")
        .load("Files/config/lynch_parameters.csv")
        .toPandas()
        .set_index("parameter")["value"]
        .to_dict()
    )
except Exception as e:
    print(f"Could not read parameter file. Using defaults. Reason: {e}")
    params = {}

PRICE_HISTORY_START = pd.to_datetime(
    params.get("price_history_start", DEFAULT_PRICE_START)
).date()

REFRESH_OVERLAP_DAYS = int(
    float(params.get("refresh_overlap_days", DEFAULT_OVERLAP_DAYS))
)

print(f"Price history start: {PRICE_HISTORY_START}")
print(f"Refresh overlap days: {REFRESH_OVERLAP_DAYS}")

# ------------------------------------------------------------
# 3. Load eligible universe from NB01
# ------------------------------------------------------------

universe = spark.sql("""
    SELECT
        symbol,
        yahoo_symbol,
        company_name,
        industry,
        rule_set
    FROM bronze_universe
    WHERE in_scope = true
""").toPandas()

if universe.empty:
    raise RuntimeError(
        "bronze_universe has zero in-scope companies. "
        "Run NB01_Universe first and review universe classification."
    )

universe["symbol"] = universe["symbol"].astype(str).str.strip().str.upper()
universe["yahoo_symbol"] = universe["yahoo_symbol"].astype(str).str.strip()

print(f"In-scope Yahoo symbols: {len(universe)}")

# ------------------------------------------------------------
# 4. Define explicit Spark schemas
# ------------------------------------------------------------

PRICE_SCHEMA = StructType([
    StructField("symbol", StringType(), False),
    StructField("yahoo_symbol", StringType(), False),
    StructField("trade_date", DateType(), False),
    StructField("open", DoubleType(), True),
    StructField("high", DoubleType(), True),
    StructField("low", DoubleType(), True),
    StructField("close", DoubleType(), True),
    StructField("adj_close", DoubleType(), True),
    StructField("volume", LongType(), True),
    StructField("ingested_at", TimestampType(), True),
    StructField("run_id", StringType(), True),
])

DIVIDEND_SCHEMA = StructType([
    StructField("symbol", StringType(), False),
    StructField("yahoo_symbol", StringType(), False),
    StructField("ex_date", DateType(), False),
    StructField("dividend", DoubleType(), True),
    StructField("ingested_at", TimestampType(), True),
    StructField("run_id", StringType(), True),
])

FUNDAMENTAL_SCHEMA = StructType([
    StructField("symbol", StringType(), False),
    StructField("yahoo_symbol", StringType(), False),
    StructField("period_type", StringType(), False),
    StructField("statement_type", StringType(), False),
    StructField("period_end", DateType(), False),
    StructField("source_field", StringType(), False),
    StructField("value", DoubleType(), True),
    StructField("ingested_at", TimestampType(), True),
    StructField("run_id", StringType(), True),
])

MARKET_SCHEMA = StructType([
    StructField("symbol", StringType(), False),
    StructField("yahoo_symbol", StringType(), False),
    StructField("company_name", StringType(), True),
    StructField("industry", StringType(), True),
    StructField("yahoo_sector", StringType(), True),
    StructField("yahoo_industry", StringType(), True),
    StructField("currency", StringType(), True),
    StructField("last_price", DoubleType(), True),
    StructField("market_cap", DoubleType(), True),
    StructField("shares_outstanding", DoubleType(), True),
    StructField("retrieved_at", TimestampType(), True),
    StructField("run_id", StringType(), True),
])

LOG_SCHEMA = StructType([
    StructField("run_id", StringType(), False),
    StructField("run_timestamp", TimestampType(), False),
    StructField("symbols_attempted", LongType(), True),
    StructField("price_success_symbols", LongType(), True),
    StructField("fundamental_success_symbols", LongType(), True),
    StructField("market_snapshot_symbols", LongType(), True),
    StructField("price_rows", LongType(), True),
    StructField("dividend_rows", LongType(), True),
    StructField("fundamental_rows", LongType(), True),
    StructField("failed_symbols", StringType(), True),
    StructField("status", StringType(), True),
])

# ------------------------------------------------------------
# 5. Determine incremental price start dates
# ------------------------------------------------------------

price_start_by_symbol = get_start_date_by_symbol(
    universe_df=universe,
    table_name="bronze_yahoo_prices",
    global_start=PRICE_HISTORY_START,
    overlap_days=REFRESH_OVERLAP_DAYS
)

# ------------------------------------------------------------
# 6. Fetch Yahoo Finance data directly
# ------------------------------------------------------------

price_rows = []
dividend_rows = []
fundamental_rows = []
market_rows = []

failed_symbols = []
price_success_symbols = set()
fundamental_success_symbols = set()

statement_sources = [
    ("income_stmt", "ANNUAL", "INCOME_STATEMENT"),
    ("balance_sheet", "ANNUAL", "BALANCE_SHEET"),
    ("cashflow", "ANNUAL", "CASH_FLOW"),
    ("quarterly_income_stmt", "QUARTERLY", "INCOME_STATEMENT"),
    ("quarterly_balance_sheet", "QUARTERLY", "BALANCE_SHEET"),
    ("quarterly_cashflow", "QUARTERLY", "CASH_FLOW"),
]

for index, row in universe.iterrows():
    symbol = str(row["symbol"])
    yahoo_symbol = str(row["yahoo_symbol"])
    start_date = price_start_by_symbol.get(symbol, PRICE_HISTORY_START)

    errors_for_symbol = []
    records_before = len(fundamental_rows)

    print(
        f"[{index + 1}/{len(universe)}] "
        f"Fetching {symbol} ({yahoo_symbol}) from {start_date}"
    )

    try:
        ticker = yf.Ticker(yahoo_symbol)

        # ----------------------------------------------------
        # A. Daily OHLCV price history
        # auto_adjust=False retains Close and Adj Close separately.
        # ----------------------------------------------------
        try:
            history = ticker.history(
                start=start_date.isoformat(),
                end=(TODAY + timedelta(days=1)).isoformat(),
                auto_adjust=False,
                actions=False
            )

            if history is not None and not history.empty:
                history = history.reset_index()

                for _, px in history.iterrows():
                    close_value = safe_float(px.get("Close"))

                    if close_value is None or close_value <= 0:
                        continue

                    trade_date = to_python_date(px.get("Date"))

                    if trade_date is None:
                        continue

                    price_rows.append((
                        symbol,
                        yahoo_symbol,
                        trade_date,
                        safe_float(px.get("Open")),
                        safe_float(px.get("High")),
                        safe_float(px.get("Low")),
                        close_value,
                        safe_float(px.get("Adj Close")),
                        int(px["Volume"])
                            if pd.notna(px.get("Volume"))
                            else None,
                        RUN_TS,
                        RUN_ID
                    ))

                price_success_symbols.add(symbol)
            else:
                errors_for_symbol.append("No price history returned")

        except Exception as e:
            errors_for_symbol.append(f"Price error: {type(e).__name__}")

        # ----------------------------------------------------
        # B. Dividend history
        # Full Yahoo dividend history is small enough to refresh.
        # ----------------------------------------------------
        try:
            dividends = ticker.dividends

            if dividends is not None and len(dividends) > 0:
                for ex_date, amount in dividends.items():
                    dividend_amount = safe_float(amount)

                    if dividend_amount is None or dividend_amount <= 0:
                        continue

                    dividend_date = to_python_date(ex_date)

                    if dividend_date is None:
                        continue

                    dividend_rows.append((
                        symbol,
                        yahoo_symbol,
                        dividend_date,
                        dividend_amount,
                        RUN_TS,
                        RUN_ID
                    ))

        except Exception as e:
            errors_for_symbol.append(f"Dividend error: {type(e).__name__}")

        # ----------------------------------------------------
        # C. Annual and quarterly statements
        # Yahoo typically exposes only recent annual/quarterly periods.
        # Store raw source rows in long format.
        # ----------------------------------------------------
        for attribute_name, period_type, statement_type in statement_sources:
            try:
                statement = getattr(ticker, attribute_name)
                flattened = flatten_statement(
                    statement_df=statement,
                    symbol=symbol,
                    yahoo_symbol=yahoo_symbol,
                    period_type=period_type,
                    statement_type=statement_type
                )
                fundamental_rows.extend(flattened)

            except Exception as e:
                errors_for_symbol.append(
                    f"{attribute_name} error: {type(e).__name__}"
                )

        if len(fundamental_rows) > records_before:
            fundamental_success_symbols.add(symbol)

        # ----------------------------------------------------
        # D. Current market metadata
        # `info` can be incomplete or rate-limited.
        # `fast_info` is used as fallback where possible.
        # ----------------------------------------------------
        yahoo_info = {}
        fast_info = None

        try:
            yahoo_info = ticker.info or {}
        except Exception as e:
            errors_for_symbol.append(f"Info error: {type(e).__name__}")

        try:
            fast_info = ticker.fast_info
        except Exception:
            fast_info = None

        market_cap = safe_float(yahoo_info.get("marketCap"))
        if market_cap is None:
            market_cap = safe_float(
                get_fast_info_value(fast_info, "market_cap")
            )

        shares = safe_float(yahoo_info.get("sharesOutstanding"))
        if shares is None:
            shares = safe_float(
                get_fast_info_value(fast_info, "shares")
            )

        last_price = safe_float(yahoo_info.get("currentPrice"))
        if last_price is None:
            last_price = safe_float(
                get_fast_info_value(fast_info, "last_price")
            )

        market_rows.append((
            symbol,
            yahoo_symbol,
            str(row["company_name"]),
            str(row["industry"]),
            str(yahoo_info.get("sector", "")),
            str(yahoo_info.get("industry", "")),
            str(yahoo_info.get("currency", "INR")),
            last_price,
            market_cap,
            shares,
            RUN_TS,
            RUN_ID
        ))

    except Exception as e:
        failed_symbols.append(symbol)
        errors_for_symbol.append(f"Ticker error: {type(e).__name__}")

    if errors_for_symbol:
        print(f"  Warnings: {' | '.join(errors_for_symbol[:3])}")

    # Be polite to Yahoo Finance.
    time.sleep(SLEEP_SECONDS)

# ------------------------------------------------------------
# 7. Write prices using Delta merge / upsert
# ------------------------------------------------------------

if len(price_rows) == 0:
    raise RuntimeError(
        "No Yahoo price rows were downloaded. "
        "Aborting to prevent overwriting historical prices with empty data."
    )

incoming_prices = (
    spark.createDataFrame(price_rows, PRICE_SCHEMA)
    .dropDuplicates(["symbol", "trade_date"])
)

if not table_exists("bronze_yahoo_prices"):
    (
        incoming_prices.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable("bronze_yahoo_prices")
    )
    print(f"Created bronze_yahoo_prices: {len(price_rows):,} input rows")
else:
    delta_prices = DeltaTable.forName(spark, "bronze_yahoo_prices")

    (
        delta_prices.alias("target")
        .merge(
            incoming_prices.alias("source"),
            "target.symbol = source.symbol "
            "AND target.trade_date = source.trade_date"
        )
        .whenMatchedUpdateAll()
        .whenNotMatchedInsertAll()
        .execute()
    )

    print(f"Merged {len(price_rows):,} price rows into bronze_yahoo_prices")

# ------------------------------------------------------------
# 8. Write full dividend history
# ------------------------------------------------------------

if len(dividend_rows) > 0:
    incoming_dividends = (
        spark.createDataFrame(dividend_rows, DIVIDEND_SCHEMA)
        .dropDuplicates(["symbol", "ex_date"])
    )

    (
        incoming_dividends.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable("bronze_yahoo_dividends")
    )

    print(f"Saved bronze_yahoo_dividends: {len(dividend_rows):,} rows")
else:
    print(
        "No dividends returned. Existing dividend table was not overwritten."
    )

# ------------------------------------------------------------
# 9. Write fundamentals safely
# ------------------------------------------------------------

minimum_success_count = max(
    1,
    int(len(universe) * MIN_FUNDAMENTAL_SUCCESS_RATIO)
)

should_write_fundamentals = (
    len(fundamental_rows) > 0
    and (
        not table_exists("bronze_yahoo_fundamentals_long")
        or len(fundamental_success_symbols) >= minimum_success_count
    )
)

if should_write_fundamentals:
    incoming_fundamentals = (
        spark.createDataFrame(fundamental_rows, FUNDAMENTAL_SCHEMA)
        .dropDuplicates([
            "symbol",
            "period_type",
            "statement_type",
            "period_end",
            "source_field"
        ])
    )

    (
        incoming_fundamentals.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable("bronze_yahoo_fundamentals_long")
    )

    print(
        f"Saved bronze_yahoo_fundamentals_long: "
        f"{len(fundamental_rows):,} rows"
    )
else:
    print(
        "Fundamental refresh incomplete. "
        "Existing bronze_yahoo_fundamentals_long was preserved."
    )

# ------------------------------------------------------------
# 10. Write market snapshot safely
# ------------------------------------------------------------

if len(market_rows) > 0:
    incoming_market = (
        spark.createDataFrame(market_rows, MARKET_SCHEMA)
        .dropDuplicates(["symbol"])
    )

    (
        incoming_market.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable("bronze_yahoo_market_snapshot")
    )

    print(f"Saved bronze_yahoo_market_snapshot: {len(market_rows):,} rows")

# ------------------------------------------------------------
# 11. Ingestion log
# ------------------------------------------------------------

status = (
    "SUCCESS"
    if len(failed_symbols) == 0
    else "PARTIAL_SUCCESS"
)

log_row = [(
    RUN_ID,
    RUN_TS,
    int(len(universe)),
    int(len(price_success_symbols)),
    int(len(fundamental_success_symbols)),
    int(len(market_rows)),
    int(len(price_rows)),
    int(len(dividend_rows)),
    int(len(fundamental_rows)),
    ",".join(failed_symbols[:100]),
    status
)]

(
    spark.createDataFrame(log_row, LOG_SCHEMA)
    .write.format("delta")
    .mode("append")
    .saveAsTable("bronze_yahoo_ingestion_log")
)

# ------------------------------------------------------------
# 12. Summary
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("YAHOO BRONZE INGESTION SUMMARY")
print("=" * 65)
print(f"Symbols attempted            : {len(universe)}")
print(f"Price-success symbols        : {len(price_success_symbols)}")
print(f"Fundamental-success symbols  : {len(fundamental_success_symbols)}")
print(f"Market snapshots             : {len(market_rows)}")
print(f"Price rows downloaded        : {len(price_rows):,}")
print(f"Dividend rows downloaded     : {len(dividend_rows):,}")
print(f"Fundamental rows downloaded  : {len(fundamental_rows):,}")
print(f"Failed ticker-level symbols  : {len(failed_symbols)}")
print(f"Run status                   : {status}")
print("=" * 65)
```

```python
%%sql
SELECT
    MIN(trade_date) AS earliest_price,
    MAX(trade_date) AS latest_price,
    COUNT(*) AS price_rows,
    COUNT(DISTINCT symbol) AS symbols
FROM bronze_yahoo_prices;
```

```python
%%sql
SELECT
    symbol,
    COUNT(DISTINCT YEAR(ex_date)) AS dividend_years,
    MIN(ex_date) AS first_dividend,
    MAX(ex_date) AS latest_dividend
FROM bronze_yahoo_dividends
GROUP BY symbol
ORDER BY dividend_years DESC;
```

```python
%%sql
SELECT
    period_type,
    statement_type,
    COUNT(*) AS rows,
    COUNT(DISTINCT symbol) AS symbols,
    MIN(period_end) AS earliest_period,
    MAX(period_end) AS latest_period
FROM bronze_yahoo_fundamentals_long
GROUP BY period_type, statement_type
ORDER BY period_type, statement_type;
```

```python
%%sql
SELECT *
FROM bronze_yahoo_ingestion_log
ORDER BY run_timestamp DESC
LIMIT 5;
```
